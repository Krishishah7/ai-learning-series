# Bag of Words (BoW)

Machine learning algorithms cannot directly understand text. Therefore, text must first be converted into numerical features.

Bag of Words (BoW) is one of the simplest and most widely used text representation techniques. It creates a vocabulary of all unique words in a collection of documents and represents each document by counting how many times each word appears.

BoW ignores grammar and word order, focusing only on the frequency of words. Although simple, it forms the foundation for many traditional NLP models.

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
documents = [
    "I love machine learning",
    "Machine learning is amazing",
    "I love artificial intelligence"
]

for doc in documents:
    print(doc)
```

```python
vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_df
```

```python
import matplotlib.pyplot as plt

word_counts = bow_df.sum(axis=0)

plt.figure(figsize=(8,4))
word_counts.plot(kind="bar")

plt.title("Word Frequencies in Corpus")
plt.xlabel("Words")
plt.ylabel("Count")

plt.show()
```

```python
binary_vectorizer = CountVectorizer(binary=True)

binary_matrix = binary_vectorizer.fit_transform(documents)

binary_df = pd.DataFrame(
    binary_matrix.toarray(),
    columns=binary_vectorizer.get_feature_names_out()
)

binary_df
```

```python
comparison = pd.DataFrame({

    "Count Representation":[
        "Stores word frequency",
        "Captures repeated words",
        "Useful for frequency analysis"
    ],

    "Binary Representation":[
        "Stores only presence/absence",
        "Ignores repetition",
        "Useful for simple text classification"
    ]

})

comparison
```

```python
steps = [
    "Raw Text",
    "Build Vocabulary",
    "Count Words",
    "Create Vector",
    "Machine Learning Model"
]

plt.figure(figsize=(10,2))

plt.plot(range(len(steps)), [1]*len(steps), marker="o")

for i, step in enumerate(steps):
    plt.text(i, 1.05, step, ha="center")

plt.xticks([])
plt.yticks([])

plt.title("Bag of Words Workflow")

plt.show()
```
