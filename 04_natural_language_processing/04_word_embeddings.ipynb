# Word Embeddings

Traditional text representation techniques such as Bag of Words (BoW) and TF-IDF represent words using sparse vectors. While effective for many classical NLP tasks, they do not capture the meaning or relationship between words.

Word Embeddings solve this problem by representing each word as a dense vector of real numbers. Words with similar meanings are placed close to each other in the vector space.

For example, words like "king" and "queen" or "car" and "vehicle" will have similar vector representations because they are semantically related.

Word embeddings form the foundation of many modern NLP models, including Word2Vec, GloVe, FastText, Transformers, and Large Language Models.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bow = pd.DataFrame({
    "artificial":[1,0],
    "machine":[1,1],
    "learning":[1,1],
    "football":[0,1]
},
index=["Document 1","Document 2"])

bow
```

```python
embeddings = {

    "king":[0.82,0.45,0.71],

    "queen":[0.80,0.47,0.69],

    "car":[0.12,0.91,0.31],

    "vehicle":[0.15,0.88,0.35]

}

embedding_df = pd.DataFrame(embeddings).T

embedding_df.columns=["Dimension 1","Dimension 2","Dimension 3"]

embedding_df
```

```python
plt.figure(figsize=(6,6))

for word, vector in embeddings.items():

    plt.scatter(vector[0], vector[1])

    plt.text(vector[0]+0.01,
             vector[1]+0.01,
             word)

plt.title("Word Embedding Space")

plt.xlabel("Dimension 1")

plt.ylabel("Dimension 2")

plt.grid(True)

plt.show()
```

```python
comparison = pd.DataFrame({

    "Bag of Words":[
        "Sparse vectors",
        "Counts words",
        "Ignores meaning"
    ],

    "TF-IDF":[
        "Sparse weighted vectors",
        "Highlights important words",
        "Limited semantics"
    ],

    "Word Embeddings":[
        "Dense vectors",
        "Captures semantic meaning",
        "Understands similarity"
    ]

})

comparison
```

```python
steps = [

    "Raw Text",

    "Vocabulary",

    "Word Embeddings",

    "Dense Vectors",

    "Deep Learning Model"

]

plt.figure(figsize=(10,2))

plt.plot(range(len(steps)),
         [1]*len(steps),
         marker="o")

for i, step in enumerate(steps):
    plt.text(i,1.05,step,ha="center")

plt.xticks([])
plt.yticks([])

plt.title("Word Embedding Workflow")

plt.show()
```
